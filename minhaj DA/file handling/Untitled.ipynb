## create and write file

```python
with open("minhaj.txt","w")as file:
    file.write("file created successfullyy!")
```

## read file

```python
with open("minhaj.txt","r")as file:
    print(file.read())
```

## overwrite file

```python
with open("minhaj.txt","w")as file:
    file.write("hello,this is file handling!")
```

## append new line

```python
with open("minhaj.txt","a")as file:
    file.write("\nAdd a new line.")
```

## read full content again 

```python
with open("minhaj.txt","r")as file:
    print(file.read())
```

## read line by line

```python
with open("minhaj.txt","r")as file:
    for line in file:
        print(line.strip())
```

## read only one line

```python
with open("minhaj.txt","r")as file:
    print(file.readline())
```

## working with file pointer

```python
with open("minhaj.txt","r")as file:
    print(file.read(10))
    file.seek(0)
    print(file.read())
    
```

## tell current position

```python
with open("minhaj.txt","r")as file:
    print(file.tell())
    file.read(9)
    print(file.tell())
```

## Read line as a list

```python
with open("minhaj.txt","r")as file:
    lines=file.readlines()
    print(lines)
```

## Write Multiple Lines

```python
lines=["Line 1\n","Line 2\n","Line 3\n"]
with open("minhaj.txt","w")as file:
    file.writelines(lines)
```

## file copy

```python
with open("minhaj.txt","r")as src:
    content=src.read()

    with open("copy.txt","w")as dest:
        dest.write(content)
```

## checking file exists


```python
import os
if os.path.exists("minhaj.txt"):
    print("file exists")
else:
    print("not exists")
```
