## Data types

```python
c=np.array([1,2,3,4,5] ,dtype=float)
print(c)
print(c.dtype)
```

```python
d=np.array([True,False,True])
d.dtype
```

## itemsize and nbytes

```python
e=np.array([2,4,6,8], dtype=np.int32)
print(e.itemsize)
print(e.nbytes)
```

```python
f=np.array([2.5,3.5,4.5,6,5], dtype=float)
print(f.itemsize)
print(f.nbytes)
```

## . Array Copy vs View

```python
g=np.array([1,2,3,4])
h=g.copy()
h[0]=900
print(g)
print(h)
```

```python
array1=np.array([10,20,30,40])
array2=array1.view()
array2[0]=100
print(array1)
print(array2)
```

##  Flatten Array

```python
b2=np.array([[10,20],
             [30,40]])
b3=b2.flatten()
b3[0]=100
print(b2)
print(b3)
```

```python
a=np.array([[1,4,5,7],
            [3,4,7,9]])
b=a.flatten()
b[4]=400
print(a)
print(b)
```

## Random module

```python
np.random.seed(0)
print(np.random.rand(5))
print(np.random.randint(2,5,10))
print(np.random.randn(4))
print(np.random.choice([40,30],5))
```

## Array iterating

```python
a=np.array([19,20,30,40])
for i in a:
    print(i)
```

```python
for i in np.nditer(a):
    print(i)
```

## joining arrays

```python
arr=np.array([10,20])
arr1=np.array([20,30])
print(np.concatenate((arr,arr1)))
```

```python
print(np.stack((arr,arr1)))
```

```python
print(np.vstack((arr,arr1)))
```

```python
print(np.hstack((arr,arr1)))
```
