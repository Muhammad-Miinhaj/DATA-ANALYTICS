```python
import numpy as np
```

```python
## Print the elements from index 1 to index 4.
a = np.array([10, 20, 30, 40, 50, 60])
print(a[1:5])
```

```python
## Find and print the data type, item size, and total memory used by the array.

b = np.array([1, 2, 3, 4], dtype=np.int32)
print(b.dtype)
print(b.itemsize)
print(b.nbytes)
```

```python
## Print its shape, then reshape it into 3 rows and 2 columns

c = np.array([[10, 20, 30], [40, 50, 60]])
print(c.shape)
print(c.reshape(3,2))
```

```python
## Create a view of the array, change the first element of the view to 100 , and
#print both the original array and the view. Check the base attribute.

a1 = np.array([[1, 2], [3, 4]])
print(a1)
a2=a1.view()
a2[0][0]=100
print(a2)
print(a2.base)
```

```python
##  Generate 5 random integers between 1 and 20 and 5 random numbers between 0 and 1.

print(np.random.randint(1,21,5))
print(np.random.rand(5))
```

```python
## Create a NumPy array containing 10, 20, 30, 40, 50 . Display the array, its
#shape, data type, size of each element, and total memory used.

a=np.array([10,20,30,40,50,])
print(a.shape)
print(a.dtype)
print(a.itemsize)
print(a.nbytes)
```

```python
##  Create a 3 × 4 array containing the numbers from 1 to 12 . Display the second and third rows.

b=np.arange(1,13).reshape(3,4)
print(b)
print("second and third row")
print(b[1],b[2])
```

```python
## Create an array containing the numbers from 1 to 12 and convert it into a 3 × 4 array

c=np.arange(1,13)
print(c.reshape(3,4))
```

```python
## Create two arrays containing [10, 20, 30] and [40, 50, 60] . Combine them so that the result contains two rows.

arr=np.array([10,20,30])
arr1=np.array([40,50,60])
result=np.vstack((arr,arr1))
print(result)
```

```python
## Divide the array [10, 20, 30, 40, 50] into three smaller arrays.

a=np.array([10,20,30,40,50])
print(np.array_split(a,3))
```

```python
## Given the array [5, 10, 15, 20, 25, 30] , display the positions of all elements greater than 15 .

b=np.array([5,10,15,20,25,30])
print(np.where(b>15))
```

```python
## Given the array [12, 5, 20, 8, 30, 15] , display only the values between 10 and 25 .

a=np.array([12,5,20,8,30,15])
b=a[(a>=10)&(a<=25)]
print(b)
```

```python
## Given the array [4, 9, 16, 25] , calculate the square root of every element.

a=np.array([4,9,16,25])
print(np.sqrt(a))
           
```

```python
#Given the array [1, 2, 3, 4] , calculate the exponential value and natural
#logarithm of every element.

a=np.array([1,2,3,4])
print(np.exp(a))
print(np.log(a))
```

```python
##  Given the array [10, 20, 30, 40, 50] , find the total, average, median,
#standard deviation, variance, smallest value, and largest value.

a=np.array([10,20,30,40,50])
print(np.sum(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.min(a))
print(np.max(a))
```

```python
## Given the array [10, 20, 5, 40, 15] , find the positions of the smallest and largest elements

a=np.array([10,20,5,40,15])
print(np.argmax(a))
print(np.argmin(a))
```

```python
## Find the sum of the values in each row and the sum of the values in each column.

a=np.array([[2,4,6],
            [8,10,12]])
print("coloumn_wise")
print(np.sum(a,axis=0))
print("row_wise")
print(np.sum(a,axis=1))
```

```python
## Create an array containing [10, 20, 30, 40] . Create a separate copy and a
#view of the array. Modify the arrays and observe the effect on the original
#array

a=np.array([10,20,30,40])
print("Original :",a)
b=a.copy()
b[1]=110
c=a.view()
c[0]=200
print("copy :",b)
print("view :",c)
```

```python
## Generate five random decimal numbers, five random integers between 1 and 20 , and five values from the list [10, 20, 30, 40] .

print(np.random.rand(5))
print(np.random.randint(1,21,5))
print(np.random.choice([10,20,30,40],5))
```

```python
## Print every element one by one, then convert the array into a one-dimensional array.

a=np.array([[1,2,3],
            [4,5,6]])
for i in np.nditer(a):
    print(i)
    
print(a.flatten())
```

```python
## Given the array [35, 10, 60, 25, 5, 45] , arrange the elements in ascending order

a=np.array([35,10,60,25,5,45])
print(np.sort(a))
```

```python
## Given the array [35, 10, 60, 25, 5, 45] , arrange the elements in descending order

a=np.array([35,10,60,25,5,45])
print(np.sort(a)[::-1])
```

```python
# Given the array ["Mango", "Apple", "Grapes", "Banana", "Orange"] , arrange the
#elements in alphabetical order.

a=np.array(["Mango","Apple","Grapes","Bnana","Orange",])
print(np.sort(a))
```

```python
## Create a NumPy array containing [10, 20, 30, 40, 50, 60] . Divide the array into four smaller arrays.

a=np.array([10,20,30,40,50,60])
print(np.array_split(a,4))
```

```python
## Create two arrays containing [1, 2, 3] and [4, 5, 6] . Combine them
#horizontally so that the result is a single row containing all six elements.

a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.hstack((a,b)))
```

```python
## Create a NumPy array containing [10, 20, 30, 40, 50] . Perform addition,
#subtraction, multiplication, division, and squaring operations on the array.

a=np.array([10,20,30,40,50])
print(a+a)
print(a-a)
print(a*a)
print(a/a)
print(a**2)
```
