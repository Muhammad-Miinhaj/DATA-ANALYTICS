```python
a=pd.Series([20,60,80,100,120])
print(a.iloc[1])
```

```python
a=pd.Series([20,60,80,100,120])
print(a.iloc[-1:])
```

```python
a=pd.Series(
    [1,2,3,4,5],
    index=["a","b","k","r","l"]
)
print(a.loc["k"])
```

```python
a=pd.Series([1,2,3,4,5])
print(a.iloc[:4])
```

```python
a=pd.Series([1,2,3,4,5])
print(a.iloc[-3:])
```

```python
a=pd.Series([1,2,3,4,5])
print("first:",a.iloc[0])
print("last:",a.iloc[-1:])
print("first three:")
print(a.iloc[:3])
print("last two:")
print(a.iloc[-2:])
```

```python
marks=pd.Series([90,80,25,65,70])
print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())
```

```python
a=pd.Series([10,20,30,40])
print(a.sum())
print(a.mean())
print(a.median())
print(a.max())
print(a.min())
```
